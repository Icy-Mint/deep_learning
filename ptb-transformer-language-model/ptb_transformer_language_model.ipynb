```python
# === Colab cell: Results Summary (SORT & BEST by Test PPL) ===
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

rows = [
    # heads, layers, train, valid(min), test, time_sec
    (8, 8, 32.00, 108.91, 104.00, 2482.8),
    (4, 8, 28.32, 109.25, 103.80, 2679.8),  # global best Test PPL
    (2, 8, 36.52, 112.37, 106.55, 2342.7),
    (8, 4, 39.31, 123.06, 115.13, 2261.5),
    (4, 4, 39.55, 124.95, 117.12, 2162.2),
    (2, 4, 42.54, 133.38, 123.26, 2153.9),
    (8, 2, 63.55, 160.37, 147.82, 1618.5),
    (4, 2, 68.15, 167.15, 153.16, 1416.4),
    (2, 2, 73.78, 173.03, 159.97, 1255.8),
]

df = pd.DataFrame(rows, columns=[
    "Heads","Layers","Train PPL","Val PPL","Test PPL","Time (s)"
])
df["Time (min)"] = (df["Time (s)"]/60).round(2)
df = df.drop(columns=["Time (s)"])

df_sorted = df.sort_values(by="Test PPL", ascending=True).reset_index(drop=True)

best_idx = df_sorted["Test PPL"].idxmin()
best_row = df_sorted.loc[best_idx]

display(Markdown("## 2.2 Results Summary (sorted by **Test PPL**)"))

def highlight_rows(row):
    return ['font-weight: 700; background-color: #f0f6ff' if row.name == best_idx else '' for _ in row]

styler = (
    df_sorted.style
    .apply(highlight_rows, axis=1)
    .set_properties(**{'text-align': 'center'})
    .set_table_styles([dict(selector='th', props=[('text-align', 'center')])])
)
styler = styler.set_properties(subset=pd.IndexSlice[best_idx, "Test PPL"], **{'font-weight': '700'})

display(styler.format({
    "Train PPL":"{:.2f}",
    "Val PPL":"{:.2f}",
    "Test PPL":"{:.2f}",
    "Time (min)":"{:.2f}",
}))
display(Markdown(
    f"**Best Configuration (by Test PPL):** {int(best_row['Heads'])} heads, {int(best_row['Layers'])} layers\n\n"
    f"- Train PPL: **{best_row['Train PPL']:.2f}**  \n"
    f"- Validation PPL: **{best_row['Val PPL']:.2f}**  \n"
    f"- **Test PPL: {best_row['Test PPL']:.2f}**  \n"
    f"- Time: {best_row['Time (min)']:.2f} min  \n"
))

fig, ax = plt.subplots(figsize=(8.6, 2 + 0.4*len(df_sorted)))
ax.axis('off')
tbl = ax.table(
    cellText=df_sorted.values,
    colLabels=df_sorted.columns,
    loc='center',
    cellLoc='center'
)
tbl.auto_set_font_size(False)
tbl.set_fontsize(10)
tbl.scale(1, 1.25)
plt.title("PTB – Results Summary (Sorted & Best by Test PPL)", pad=12)
plt.show()
```
